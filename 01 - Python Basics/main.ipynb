| Method                             | Description                                       |
| ---------------------------------- | ------------------------------------------------- |
| `append(x)`                        | Add an item to the end                            |
| `extend(iterable)`                 | Add all items from another iterable               |
| `insert(i, x)`                     | Insert at position `i`                            |
| `remove(x)`                        | Remove the first matching value                   |
| `pop([i])`                         | Remove and return item at index (last by default) |
| `clear()`                          | Remove all items                                  |
| `index(x[, start[, end]])`         | Return first index of value                       |
| `count(x)`                         | Count how many times `x` appears                  |
| `sort(*, key=None, reverse=False)` | Sort the list in-place                            |
| `reverse()`                        | Reverse the list in-place                         |
| `copy()`                           | Return a shallow copy                             |


```python
# Initial list
fruits = ["apple", "banana"]
print("Original list:", fruits)

# 1. append() - Add one item to the end
fruits.append("mango")
print("After append:", fruits)

# 2. extend() - Add multiple items
fruits.extend(["orange", "grape"])
print("After extend:", fruits)

# 3. insert() - Insert at a specific index
fruits.insert(1, "kiwi")
print("After insert at index 1:", fruits)

# 4. remove() - Remove the first matching item
fruits.remove("banana")
print("After removing 'banana':", fruits)

# 5. pop() - Remove and return the last item
popped_item = fruits.pop()
print("After pop (removed last item):", fruits)
print("Popped item:", popped_item)

# 6. clear() - Remove all items (commented out so we can continue using the list)
# fruits.clear()
# print("After clear:", fruits)

# 7. index() - Find the index of an item
apple_index = fruits.index("apple")
print("Index of 'apple':", apple_index)

# 8. count() - Count occurrences of an item
fruits.append("apple")
apple_count = fruits.count("apple")
print("After appending another 'apple':", fruits)
print("Count of 'apple':", apple_count)

# 9. sort() - Sort the list
fruits.sort()
print("After sort:", fruits)

# 10. reverse() - Reverse the list
fruits.reverse()
print("After reverse:", fruits)

# 11. copy() - Make a shallow copy
fruits_copy = fruits.copy()
print("Copy of the list:", fruits_copy)

# --- Bonus Built-in Functions ---
print("Length of list:", len(fruits))
print("Max item:", max(fruits))   # Alphabetically
print("Min item:", min(fruits))   # Alphabetically

# For numeric list (example)
numbers = [5, 2, 8, 1]
print("\nNumbers list:", numbers)
print("Sum of numbers:", sum(numbers))
print("Sorted numbers:", sorted(numbers))
```

Dictionary

```python
# 1. Creating a dictionary
person = {
    "name": "Alice",
    "age": 25,
    "city": "Lahore"
}
print("Original dictionary:", person)

# 2. Accessing values
print("Name:", person["name"])  # Alice

# 3. Using get() to access value safely
print("Age:", person.get("age"))         # 25
print("Country (with default):", person.get("country", "Pakistan"))  # Pakistan

# 4. Changing values
person["age"] = 30
print("Updated age:", person)

# 5. Adding new key-value pair
person["email"] = "alice@example.com"
print("After adding email:", person)

# 6. Removing a key using pop()
person.pop("city")
print("After pop:", person)

# 7. Removing the last inserted item (Python 3.7+)
last_item = person.popitem()
print("Popped last item:", last_item)
print("After popitem:", person)

# 8. Check if key exists
if "name" in person:
    print("Yes, 'name' is a key.")

# 9. Loop through keys
print("Keys:")
for key in person:
    print(key)

# 10. Loop through values
print("Values:")
for value in person.values():
    print(value)

# 11. Loop through items
print("Items (key and value):")
for key, value in person.items():
    print(key, ":", value)

# 12. Length of dictionary
print("Number of items:", len(person))

# 13. Copy a dictionary
new_person = person.copy()
print("Copied dictionary:", new_person)

# 14. Clear the dictionary
person.clear()
print("After clear:", person)

# 15. Create using dict() constructor
info = dict(name="Bob", age=22, city="Karachi")
print("Created using dict():", info)

# 16. Nested dictionary
students = {
    "student1": {"name": "Ali", "age": 20},
    "student2": {"name": "Zara", "age": 21}
}
print("Nested dictionary:", students)
print("Student1's name:", students["student1"]["name"])
```

set

```python
# 1. Creating a set
fruits = {"apple", "banana", "mango"}
print("Original set:", fruits)

# 2. Adding an item
fruits.add("orange")
print("After adding 'orange':", fruits)

# 3. Adding multiple items using update()
fruits.update(["grape", "kiwi"])
print("After update:", fruits)

# 4. Removing an item using remove()
fruits.remove("banana")
print("After removing 'banana':", fruits)

# 5. Discarding an item (no error if not found)
fruits.discard("papaya")  # No error even if it doesn’t exist
print("After discard (non-existing):", fruits)

# 6. pop() – removes a random item
removed = fruits.pop()
print("Item removed with pop():", removed)
print("After pop:", fruits)

# 7. Clear the set
temp = fruits.copy()
temp.clear()
print("After clear:", temp)

# 8. Copy a set
copy_set = fruits.copy()
print("Copied set:", copy_set)

# 9. Length of set
print("Length of set:", len(fruits))

# 10. Membership test
print("Is 'apple' in set?", "apple" in fruits)

# 11. Set operations
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

# Union (all items)
print("Union:", set1.union(set2))         # or set1 | set2

# Intersection (common items)
print("Intersection:", set1.intersection(set2))  # or set1 & set2

# Difference (items in set1 not in set2)
print("Difference (set1 - set2):", set1.difference(set2))  # or set1 - set2

# Symmetric Difference (items in either but not both)
print("Symmetric Difference:", set1.symmetric_difference(set2))  # or set1 ^ set2

# 12. isdisjoint(), issubset(), issuperset()
print("Is disjoint:", set1.isdisjoint({7, 8}))
print("Is subset:", {1, 2}.issubset(set1))
print("Is superset:", set1.issuperset({1, 2}))
```

Iterators vs Generators

Iterators do not store all elements in memory at once

```python
nums = iter([1, 2, 3])
print(next(nums))  # 1
```

```python
my_list = [1, 2, 3]
iterator = iter(my_list)  # Convert list to iterator

# Loop through the iterator
for item in iterator:
    print(item)  # Automatically stops when StopIteration is raised
```

Generator

- A generator is a simpler and cleaner way to create an iterator using the yield keyword.
- Automatically creates the __iter__() and __next__() for you.
- A special function that yields values one by one.
- Automatically saves state between each call
- Generator saves memory (lazy evaluation)






```python
def my_gen():
    for i in range(1, 4):
        yield i

gen = my_gen()
print(next(gen))  # 1
print(next(gen))  # 2
print(next(gen))  # 3
# next(gen) again → StopIteration
```

```python
def big_range():
    for i in range(1, 1000000):
        yield i

gen = big_range()

for i in gen:
    if i > 10:
        break  # Stop after 10 iterations
    print(i)
```

 List Comprehension

[expression for item in iterable if condition]


- expression: what you want to include in the list
- item: current value from the iterable
- condition (optional): filter to include only certain items

```python
squares = [x*x for x in range(2,6)] 
```

```python
squares
```

```python
even = [x for x in range(10) if x % 2 == 0]
print(even)
```

```python
labels = ["even" if x % 2 == 0 else "odd" for x in range(5)]
print(labels)
# ['even', 'odd', 'even', 'odd', 'even']
```

Error Handling – try and except

| Error Type          | When it happens                          |
| ------------------- | ---------------------------------------- |
| `ZeroDivisionError` | Dividing by 0                            |
| `ValueError`        | Wrong type of value (e.g., `int("abc")`) |
| `TypeError`         | Wrong type used (e.g., adding str + int) |
| `IndexError`        | List index out of range                  |
| `KeyError`          | Dictionary key not found                 |
| `FileNotFoundError` | File does not exist                      |


```python
try:
    x = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero!")
```

```python
try:
    age = int(input("Enter your age: "))
    print("You are", age, "years old.")
except ValueError:
    print("Please enter a valid number.")
```

```python
try:
    x = int("abc")
    y = 10 / 0
except ValueError:
    print("Value Error occurred")
except ZeroDivisionError:
    print("Division by zero is not allowed")
```

```python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    result = a / b
except ZeroDivisionError:
    print("You cannot divide by zero!")
except ValueError:
    print("Please enter valid numbers!")
else:
    print("Result is:", result)
finally:
    print("Done!")
```
